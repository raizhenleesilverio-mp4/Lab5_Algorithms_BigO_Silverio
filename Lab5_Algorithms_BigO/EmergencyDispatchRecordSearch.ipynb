```python
# [CELL 1] PYTHON RUNTIME TEST
import sys
import platform
print("Python Version:", sys.version.split()[0])
print("Operating System:", platform.system())
print("Python Executable:", sys.executable)
```

```python
# [CELL 2] STUDENT CONFIGURATION (student-specific input rules)
LAST_NAME = "Silverio"
STUDENT_ID = "TUPM-26-1261"
FAVORITE_ARTIST = "YAELOKRE" 

SEED_DIGIT = int(STUDENT_ID[-1])                                  # last digit of ID
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())          # sum of ID digits
NAME_LENGTH = len(LAST_NAME)                                      # length of surname

_artist_letters = [c for c in FAVORITE_ARTIST.upper() if c.isalpha()]
ARTIST_LENGTH = len(_artist_letters)                              # letters only (spaces ignored)
ARTIST_SUM = sum(ord(c) - ord("A") + 1 for c in _artist_letters)  # A=1 ... Z=26

print("LAST_NAME      :", LAST_NAME)
print("STUDENT_ID     :", STUDENT_ID)
print("FAVORITE_ARTIST:", FAVORITE_ARTIST)
print("SEED_DIGIT     :", SEED_DIGIT)
print("ID_SUM         :", ID_SUM)
print("NAME_LENGTH    :", NAME_LENGTH)
print("ARTIST_LENGTH  :", ARTIST_LENGTH)
print("ARTIST_SUM     :", ARTIST_SUM)
```

```python
# [CELL 3] GENERATE ORIGINAL AND ROTATED INCIDENT RECORDS
n1 = 20 + ID_SUM                       # number of incident records
base = 1000 + ID_SUM                   # first incident number
step = NAME_LENGTH + 1                 # gap between incidents (keeps values unique)

original_records = [base + i * step for i in range(n1)]   # numerical order

rotation_point = SEED_DIGIT % n1       # records moved from the beginning to the end
rotated_records = original_records[rotation_point:] + original_records[:rotation_point]

target_incident = original_records[(ID_SUM + ARTIST_SUM) % n1]   # target (present)
missing_incident = target_incident + 1                            # target (absent)

print("Original records :", original_records)
print("Rotation point  :", rotation_point)
print("Rotated records  :", rotated_records)
print("Target incident  :", target_incident)
print("Missing incident :", missing_incident)
```

```python
# [CELL 4] LINEAR SEARCH
def linear_search(data, target):
    comparisons = 0
    for i, value in enumerate(data):
        comparisons += 1
        if value == target:
            return i, comparisons
    return -1, comparisons
```

```python
# [CELL 5] ROTATED BINARY SEARCH  (O(log n))
# Every element comparison is counted: equality check, sorted-half check,
# and range check.
def rotated_binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons

        comparisons += 1
        if data[low] <= data[mid]:                 # left half is sorted
            comparisons += 1
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:                                      # right half is sorted
            comparisons += 1
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1, comparisons
```

```python
# [CELL 6] SEARCH RESULTS AND COMPARISON WITH LINEAR SEARCH
def report(label, target):
    li, lc = linear_search(rotated_records, target)
    bi, bc = rotated_binary_search(rotated_records, target)
    print(f"\n--- {label}: {target} ---")
    for name, idx, cnt in (("Linear Search", li, lc), ("Rotated Binary Search", bi, bc)):
        status = f"found at index {idx}" if idx != -1 else "NOT FOUND"
        print(f"{name:<22}: {status:<18} comparisons={cnt}")
    print("Comparison reduction:", lc - bc)
    return lc, bc

ex1_found = report("Target (present)", target_incident)
ex1_missing = report("Target (absent)", missing_incident)
```
